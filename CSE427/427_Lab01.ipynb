## Part 1: NumPy

```python
import numpy as np
```

```python
# Task 1: Create two linear arrays a and b with 20 random elements each ranging from 1 to 100
a = np.random.randint(1, 101, 20)
b = np.random.randint(1, 101, 20)
print("Array a:", a)
print("Array b:", b)
```

```python
# Task 2: Reshape linear arrays into 5x4 matrices
a = a.reshape(5, 4)
b = b.reshape(5, 4)
print("Matrix a (5x4):")
print(a)
print("\nMatrix b (5x4):")
print(b)
```

```python
# Task 3: Perform scalar multiplication with matrix a and b (element-wise)
scalar_mult = a * b
print("Scalar (element-wise) multiplication of a and b:")
print(scalar_mult)
```

```python
# Task 4: Modify matrix b (transpose it to 4x5) and perform matrix multiplication a @ b to store in c
# For matrix multiplication a (5x4) @ b_T (4x5) -> c (5x5)
b = b.T  # modify b to 4x5 by transposing
print("Modified matrix b (transposed to 4x5):")
print(b)

c = np.matmul(a, b)  # or a @ b
print("\nMatrix c = a @ b (5x5):")
print(c)
```

```python
# Task 5: Print only column 2 and 3 of matrix c; print the whole matrix using slicing
print("Columns 2 and 3 of matrix c:")
print(c[:, 2:4])   # columns at index 2 and 3

print("\nWhole matrix c using slicing:")
print(c[:, :])     # all rows, all columns
```

```python
# Task 6: Find the largest element in c and its index
largest = np.max(c)
index_flat = np.argmax(c)           # flat index
index_2d = np.unravel_index(index_flat, c.shape)  # (row, col)

print("Largest element in c:", largest)
print("Flat index of largest element:", index_flat)
print("2D index (row, col) of largest element:", index_2d)
```

```python
# Task 7: Convert matrix c into a 1D array without changing the original
c_flat = c.flatten()   # flatten() returns a copy, original c is unchanged
print("1D version of c:")
print(c_flat)
print("\nOriginal matrix c (unchanged):")
print(c)
```

## Part 2: Pandas

```python
import pandas as pd
```

```python
# Task 1: Read the CSV file as a DataFrame
# Download from Google Drive (using a direct download link)
url = "https://drive.google.com/uc?id=1zo5y462G1FaGUlAEUayACHSV2zSliR-5"
df = pd.read_csv(url)
print("CSV file loaded successfully.")
```

```python
# Task 2: Show the first few rows of the DataFrame
df.head()
```

```python
# Task 3: Show summarized details of the dataset
print("--- DataFrame Info ---")
df.info()
print("\n--- Statistical Summary ---")
df.describe()
```

```python
# Task 4: Show how many null values are in each column
print("Null value count per column:")
print(df.isnull().sum())
```

```python
# Task 5: Replace null values in each column with the mean of that column
for col in df.select_dtypes(include='number').columns:
    df[col] = df[col].fillna(df[col].mean())

print("Null values after replacement:")
print(df.isnull().sum())
```

## Part 3: Matplotlib

```python
import matplotlib.pyplot as plt

temperature_dhaka = [25, 34, 21, 45, 28, 6, 43, 18, 7, 2]
humidity_dhaka    = [28, 25, 29, 20, 26, 50, 19, 29, 52, 55]

study_hours = [2, 3, 4, 4, 5, 6, 7, 7, 8, 9, 9, 10, 11, 11, 12]
marks       = [6, 10, 15, 20, 34, 44, 55, 60, 55, 67, 70, 80, 90, 99, 100]

subjects      = ['Maths', 'English', 'Science', 'Physics', 'Computer']
subject_marks = [89, 90, 45, 78, 99]
```

```python
# Task 1: Scatter plot — temperature_dhaka vs humidity_dhaka, marker='*', color=red
plt.scatter(temperature_dhaka, humidity_dhaka, marker='*', color='red')
plt.title('Temperature vs Humidity (Dhaka)')
plt.xlabel('Temperature (°C)')
plt.ylabel('Humidity (%)')
plt.grid(True)
plt.show()
```

```python
# Task 2: Line plot — study_hours vs marks, figure size (12, 8)
plt.figure(figsize=(12, 8))
plt.plot(study_hours, marks)
plt.title('Study Hours vs Marks')
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.grid(True)
plt.show()
```

```python
# Task 3: Horizontal bar graph — subject scores
plt.barh(subjects, subject_marks)
plt.title('Subject Scores (Horizontal Bar)')
plt.xlabel('Marks')
plt.ylabel('Subjects')
plt.show()
```

```python
# Task 4: Vertical bar graph — each subject in a different color
colors = ['red', 'blue', 'green', 'orange', 'purple']
plt.bar(subjects, subject_marks, color=colors)
plt.title('Subject Scores (Vertical Bar)')
plt.xlabel('Subjects')
plt.ylabel('Marks')
plt.show()
```

```python
# Task 5: All 4 graphs in one figure using subplot()
plt.figure(figsize=(14, 10))

# Subplot 1 — Scatter plot
plt.subplot(2, 2, 1)
plt.scatter(temperature_dhaka, humidity_dhaka, marker='*', color='red')
plt.title('Temperature vs Humidity')
plt.xlabel('Temperature (°C)')
plt.ylabel('Humidity (%)')
plt.grid(True)

# Subplot 2 — Line plot
plt.subplot(2, 2, 2)
plt.plot(study_hours, marks)
plt.title('Study Hours vs Marks')
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.grid(True)

# Subplot 3 — Horizontal bar
plt.subplot(2, 2, 3)
plt.barh(subjects, subject_marks)
plt.title('Subject Scores (Horizontal)')
plt.xlabel('Marks')

# Subplot 4 — Vertical bar with colors
plt.subplot(2, 2, 4)
plt.bar(subjects, subject_marks, color=colors)
plt.title('Subject Scores (Vertical)')
plt.xlabel('Subjects')
plt.ylabel('Marks')

plt.tight_layout()
plt.show()
```
